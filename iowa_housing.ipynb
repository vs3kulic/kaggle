# Introduction

In this exercise, we create and submit predictions for a Kaggle competition. We start by importing essential libraries and setting the filepaths for the dataset.

````python
import os
import pandas as pd
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error
from sklearn.model_selection import train_test_split
````

We repeat the known steps: 

1. Load the data
2. Separate the target variable
3. Select the features to use
4. Split into validation and training data
5. Define and fit a model
6. Make predictions on validation data
7. Calculate the MAE

````python
# Load the data, and separate the target variable
iowa_file_path = "datasets/iowa-housing/train.csv"
home_data = pd.read_csv(iowa_file_path)
y = home_data.SalePrice

# Select the features to use
features = ['LotArea', 'YearBuilt', '1stFlrSF', '2ndFlrSF', 'FullBath', 'BedroomAbvGr', 'TotRmsAbvGrd']

# Select columns corresponding to features, and preview the data
X = home_data[features]
X.head()

# Split into validation and training data
train_X, val_X, train_y, val_y = train_test_split(X, y, random_state = 1)

# Define and fit a random forest model
rf_model = RandomForestRegressor(random_state = 1)
rf_model.fit(train_X, train_y)

# Make predictions on validation data, and compute the MAE
rf_val_preds = rf_model.predict(val_X)
rf_val_mae = mean_absolute_error(rf_val_preds, val_y)

print("Validation MAE for Random Forest Model: {:,.0f}".format(rf_val_mae))
````

# Train a model for the competition

The code cell above trains a Random Forest model on **`train_X`** and **`train_y`**.  

We use the code cell below to build a Random Forest model and train it on all of **`X`** and **`y`**.

````python
# To improve accuracy, create a new Random Forest model which you will train on all training data
rf_model_on_full_data = RandomForestRegressor(random_state = 1)

# fit rf_model_on_full_data on all data from the training data
rf_model_on_full_data.fit(X, y)
````

Now, we read the "test" data and apply our model to make predictions.

````python
# Path to file you will use for predictions
test_data_path = 'datasets/iowa-housing/test.csv'

# Read test data file using pandas
test_data = pd.read_csv(test_data_path)

# Create test_X which comes from test_data but includes only the columns you used for prediction.
test_X = test_data[features]

# Make predictions which we will submit
test_preds = rf_model_on_full_data.predict(test_X)
````

Before submitting, run a check to make sure your `test_preds` have the right format.

## Model Evaluation

We can only evaluate our model's performance on the **validation set** locally. The validation MAE is approximately **{rf_val_mae}**.

To see how well the model performs on the **test set**, we must submit our predictions to Kaggle. The competition leaderboard will show our final score based on the hidden test labels.

# Generate a submission

Run the code cell below to generate a CSV file with your predictions that you can use to submit to the competition.

````python
# Run the code to save predictions in the format used for competition scoring

output = pd.DataFrame({'Id': test_data.Id,
                       'SalePrice': test_preds})
output.to_csv('submission.csv', index=False)
````
